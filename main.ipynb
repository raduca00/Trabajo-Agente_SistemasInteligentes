```python
from datetime import datetime, timedelta

class SmartStudyAgent:
    def __init__(self):
        # Componente 4: Memoria volátil de trabajo (lista en ejecución)
        self.tareas = []
        print("🤖 SmartStudy Agent: ¡Hola! Entorno agéntico inicializado.")

    # Componente 1: Analizador e introductor manual de tareas
    def menu_introducir_tareas(self):
        print("\n📥 --- Modo: Introducir Tareas (Escribe 'exit' en el nombre para terminar) ---")
        
        while True:
            nombre = input("\n📝 Nombre de la tarea/examen: ").strip()
            if nombre.lower() == 'exit':
                print("🚪 Saliendo del modo inserción de tareas.")
                break
                
            if not nombre:
                print("⚠️ El nombre no puede estar vacío.")
                continue

            # Validación de días para la entrega
            while True:
                try:
                    dias_input = input(f"📅 ¿En cuántos días se entrega '{nombre}'?: ")
                    if dias_input.lower() == 'exit': return
                    dias_para_entrega = int(dias_input)
                    if dias_para_entrega < 0:
                        print("⚠️ Los días no pueden ser negativos.")
                        continue
                    break
                except ValueError:
                    print("⚠️ Por favor, introduce un número entero válido.")

            # Validación de horas estimadas
            while True:
                try:
                    horas_input = input(f"⏳ ¿Cuántas horas de estudio/trabajo estimas que requiere?: ")
                    if horas_input.lower() == 'exit': return
                    horas_estimadas = float(horas_input)
                    if horas_estimadas <= 0:
                        print("⚠️ Las horas deben ser mayores que cero.")
                        continue
                    break
                except ValueError:
                    print("⚠️ Por favor, introduce un número válido (puede tener decimales).")

            # Estructura interna de la tarea
            fecha_limite = datetime.now() + timedelta(days=dias_para_entrega)
            tarea = {
                "nombre": nombre,
                "dias_para_entrega": dias_para_entrega,
                "fecha_limite": fecha_limite,
                "horas_estimadas": horas_estimadas,
                "prioridad": 0.0
            }
            
            # Se añade al final de las previas (funciona como un append acumulativo)
            self.tareas.append(tarea)
            print(f"✅ ¡Añadida! Tareas en memoria actual: {len(self.tareas)}")

    # Componente 2: Priorizador (Razonamiento basado en reglas)
    def calcular_prioridades(self):
        if not self.tareas:
            return
        print("\n🧠 Pensando... Calculando prioridades de las tareas...")
        for tarea in self.tareas:
            dias = max(tarea["dias_para_entrega"], 0.5) 
            tarea["prioridad"] = tarea["horas_estimadas"] / dias
        
        # Ordenamos de mayor a menor urgencia
        self.tareas.sort(key=lambda x: x["prioridad"], reverse=True)

    # Componente 3: Planificador (Toma de decisiones)
    def generar_plan(self):
        if not self.tareas:
            print("\n⚠️ No hay tareas registradas en la memoria del agente. ¡Añade algunas primero!")
            return

        while True:
            try:
                horas_input = input("\n📅 ¿De cuántas horas dispones al día para estudiar?: ")
                horas_disponibles_al_dia = float(horas_input)
                if horas_disponibles_al_dia <= 0:
                    print("⚠️ Debes disponer de al menos algo de tiempo al día.")
                    continue
                break
            except ValueError:
                print("⚠️ Introduce un número válido.")

        print(f"\n📅 --- CALENDARIO DE ESTUDIO GENERADO ({horas_disponibles_al_dia}h/día) ---")
        self.calcular_prioridades()
        
        dia_actual = datetime.now()
        tareas_pendientes = [t.copy() for t in self.tareas]
        
        # Planificamos los próximos 5 días
        for i in range(5):
            fecha_plan = dia_actual + timedelta(days=i)
            nombre_dia = fecha_plan.strftime('%A')
            
            horas_restantes_hoy = horas_disponibles_al_dia
            bloques_hoy = []
            
            for tarea in tareas_pendientes:
                if horas_restantes_hoy <= 0:
                    break
                
                if tarea["horas_estimadas"] > 0:
                    horas_a_dedicar = min(tarea["horas_estimadas"], horas_restantes_hoy)
                    tarea["horas_estimadas"] -= horas_a_dedicar
                    horas_restantes_hoy -= horas_a_dedicar
                    bloques_hoy.append(f"{tarea['nombre']} ({horas_a_dedicar}h)")
            
            if bloques_hoy:
                print(f"  • {nombre_dia}: {", ".join(bloques_hoy)}")
            else:
                print(f"  • {nombre_dia}: ☕ ¡Libre o sin tareas pendientes!")

    # Funciones auxiliares de gestión de memoria solicitadas
    def listar_tareas(self):
        if not self.tareas:
            print("\n📭 La memoria está vacía. No hay tareas.")
            return
        print("\n📋 --- Listado de tareas actuales en memoria ---")
        for i, t in enumerate(self.tareas, 1):
            print(f"  {i}. {t['nombre']} | Entrega en: {t['dias_para_entrega']} días | Horas necesarias: {t['horas_estimadas']}h")

    def borrar_tareas(self):
        self.tareas = []
        print("\n🗑️ Memoria volátil limpiada con éxito. Se han eliminado todas las tareas.")

# --- BUCLE PRINCIPAL DE INTERACCIÓN (MENÚ) ---
def iniciar_interfaz():
    agente = SmartStudyAgent()
    
    while True:
        print("\n========================================")
        print("🤖 MENÚ DEL AGENTE SMARTSTUDY")
        print("========================================")
        print("1. 📥 Introducir tareas manualmente")
        print("2. 📋 Ver tareas actuales en memoria")
        print("3. 📅 Generar calendario de estudio")
        print("4. 🗑️ Borrar tareas de la memoria")
        print("5. ❌ Salir del programa")
        print("========================================")
        
        opcion = input("Selecciona una opción (1-5): ").strip()
        
        if opcion == "1":
            agente.menu_introducir_tareas()
        elif opcion == "2":
            agente.listar_tareas()
        elif opcion == "3":
            agente.generar_plan()
        elif opcion == "4":
            agente.borrar_tareas()
        elif opcion == "5":
            print("\n👋 ¡Hasta luego! Guardando sesión mental... Apagando agente.")
            break
        else:
            print("\n⚠️ Opción no válida. Por favor, introduce un número del 1 al 5.")

# Ejecutar el menú interactivo
iniciar_interfaz()
```
